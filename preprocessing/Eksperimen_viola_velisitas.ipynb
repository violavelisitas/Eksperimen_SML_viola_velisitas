## 1. Import Library

```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.ensemble import RandomForestClassifier
import warnings
warnings.filterwarnings('ignore')
print('Library berhasil diimport')
```

## 2. Data Loading

```python
df = pd.read_csv('../heart_raw/heart_raw.csv')
print('Shape:', df.shape)
df.head()
```

## 4. Data Preprocessing

```python
# Hapus duplikat
df = df.drop_duplicates()
print('Shape setelah hapus duplikat:', df.shape)
```

```python
# Pisahkan fitur dan target
X = df.drop('target', axis=1)
y = df['target']
print('X shape:', X.shape)
print('y shape:', y.shape)
```

```python
# Normalisasi dengan StandardScaler
scaler = StandardScaler()
X_scaled = pd.DataFrame(scaler.fit_transform(X), columns=X.columns)
print('Scaling selesai')
print(X_scaled.describe().round(2).loc[['mean','std']])
```

```python
# Train-test split 80:20
X_train, X_test, y_train, y_test = train_test_split(
    X_scaled, y, test_size=0.2, random_state=42
)
print(f'Train: {X_train.shape}')
print(f'Test:  {X_test.shape}')
```

## 5. Simpan Dataset Preprocessing

```python
import os
os.makedirs('../heart_preprocessing', exist_ok=True)
df_save = X_scaled.copy()
df_save['target'] = y.values
df_save.to_csv('../heart_preprocessing/heart_preprocessing.csv', index=False)
print('Tersimpan! Shape:', df_save.shape)
```
